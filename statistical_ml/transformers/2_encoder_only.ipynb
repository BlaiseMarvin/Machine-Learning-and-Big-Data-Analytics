**BERT**
- Pretrained on both MLM (Masked Language Modelling) and NSP (Next Sentence Prediction) tasks. 
- CLS token at the start carries the embedding representation of the entire input sentence, however it was found out that you get better sentence embeddings by pooling the contextualised embeddings.
